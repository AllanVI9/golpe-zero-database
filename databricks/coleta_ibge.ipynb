```python

import requests

url = "https://apisidra.ibge.gov.br/values/t/9514/n3/35/v/93/p/2022/c2/6794/c286/113635/c287/93095,93096,93097,93098,49108,49109,60040,60041,6653"

# Coleta da API
resposta = requests.get(url, timeout=30)
resposta.raise_for_status()

# Converte JSON
dados_json = resposta.json()

# Remove a primeira linha, que contém os cabeçalhos
registros = dados_json[1:]

# Organiza os dados
dados = []

for item in registros:
    dados.append((
        item["D1N"],
        int(item["D3N"]),
        item["D6N"],
        int(item["V"])
    ))

# Cria DataFrame
df = spark.createDataFrame(
    dados,
    ["estado", "ano", "faixa_etaria", "populacao"]
)

# Salva na Bronze
df.write.mode("overwrite").saveAsTable(
    "golpe_zero.bronze_ibge"
)

print("Dados do IBGE carregados com sucesso!")
print("Total de registros:", df.count())

display(df)

```
